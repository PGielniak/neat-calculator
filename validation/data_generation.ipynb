```python
import numpy as np
import pandas as pd

source_path = "merged_labeled_data.csv"
output_path = "merged_labeled_data_noisy.csv"

df = pd.read_csv(source_path)
label_col = "label"
timestamp_cols = ["timestamp", "timestampNanos"]

noise_cols = [c for c in df.columns if c not in ([label_col] + timestamp_cols)]

rng = np.random.default_rng(42)
stds = df[noise_cols].std()
stds = stds.replace(0, np.nan)
noise = rng.normal(loc=0.0, scale=stds * 0.02, size=(len(df), len(noise_cols)))
noise_df = pd.DataFrame(noise, columns=noise_cols)
noisy_df = df.copy()

# drop all rows where label is not in the list of approved labels
approved_labels = ['LAYING', 'SITTING', 'STANDING', 'WALKING', 'WALKING_DOWNSTAIRS', 'WALKING_UPSTAIRS']

noisy_df = noisy_df[noisy_df[label_col].isin(approved_labels)].reset_index(drop=True)


noisy_df[noise_cols] = noisy_df[noise_cols] + noise_df
noisy_df.to_csv(output_path, index=False)

print(f"Wrote noisy data to {output_path}")
```
